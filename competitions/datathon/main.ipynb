```python
import pandas as pd
from pathlib import Path
 
path = Path("./data")

train_df = pd.read_csv(path / "train.csv")
test_df = pd.read_csv(path / "test.csv")
```

```python
train_df.groupby("user_session").nunique()
```

```python
train_df["event_type"].value_counts()
```

```python
train_df["event_time"] = pd.to_datetime(train_df["event_time"])
```

```python
from sklearn.preprocessing import OrdinalEncoder

oe = OrdinalEncoder()

train_df["event_type"] = oe.fit_transform(train_df[["event_type"]]).astype(int)
test_df["event_type"] = oe.transform(test_df[["event_type"]]).astype(int)
```

```python
train_df["product_id"] = train_df["product_id"].str.extract(r'(\d+)$').astype("Int64")
train_df["user_id"]    = train_df["user_id"].str.extract(r'(\d+)$').astype("Int64")
train_df["category_id"] = train_df["category_id"].str.extract(r'(\d+)$').astype("Int64")
train_df["user_session"] = train_df["user_session"].str.extract(r'(\d+)$').astype("Int64")
```

```python
import numpy as np
train_df["session_value"] = np.log1p(train_df["session_value"])
```

```python
test_df["product_id"] = test_df["product_id"].str.extract(r'(\d+)$').astype("Int64")
test_df["user_id"]    = test_df["user_id"].str.extract(r'(\d+)$').astype("Int64")
test_df["category_id"] = test_df["category_id"].str.extract(r'(\d+)$').astype("Int64")
test_df["user_session"] = test_df["user_session"].str.extract(r'(\d+)$').astype("Int64")
```

```python
train_df.to_csv("train_normalize.csv")
test_df.to_csv("test_normalize.csv")
```
