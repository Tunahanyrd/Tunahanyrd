```python
from pathlib import Path
import pandas as pd
import numpy as np
path = Path("data")
```

```python
train_df = pd.read_csv(path / "train.csv", index_col="Id")
test_df = pd.read_csv(path / "test.csv", index_col="Id")
```

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import StandardScaler, PolynomialFeatures, OrdinalEncoder
from sklearn.feature_selection import VarianceThreshold
from sklearn.linear_model import RidgeCV, LassoCV, ElasticNetCV
from sklearn.ensemble import ExtraTreesRegressor
from category_encoders import TargetEncoder, CountEncoder
import lightgbm as lgb
import xgboost as xgb
from catboost import CatBoostRegressor
from scipy.optimize import minimize
```

```python
def safe_div(a, b):
    a = a.astype(float)
    b = b.astype(float)
    out = np.divide(a, b, out=np.zeros_like(a, dtype=float), where=(b != 0))
    out[~np.isfinite(out)] = 0.0
    return out

def preprocessing(X_tr, X_va, X_te, y_tr):
    X_tr = X_tr.copy()
    X_va = X_va.copy()
    X_te = X_te.copy()

    def apply_custom_mappings(df):
        df["Condition1"] = df["Condition1"].apply(lambda x: 0 if x == "Norm" else 1)
        df["MSZoning"]   = df["MSZoning"].apply(lambda x: 0 if x == "RL" else (1 if x == "RM" else 2))
        df["CentralAir"] = df["CentralAir"].apply(lambda x: 1 if x == "Y" else 0)
        df["Functional"] = df["Functional"].apply(lambda x: 0 if x == "Typ" else 1)
        return df
    
    for df_ in (X_tr, X_va, X_te):
        df_[["Condition1","MSZoning","CentralAir","Functional"]] = apply_custom_mappings(
            df_[["Condition1","MSZoning","CentralAir","Functional"]]
        )
    gyb_max = X_tr["GarageYrBlt"].max()
    for df_ in (X_tr, X_va, X_te):
        df_["GarageYrBlt"] = gyb_max - df_["GarageYrBlt"]
        df_["GarageYrBlt"] = df_["GarageYrBlt"].fillna(-1)

    for df_ in (X_tr, X_va, X_te):
        df_["GarageType"] = df_["GarageType"].fillna("Nogrg")

    for df_ in (X_tr, X_va, X_te):
        df_["BsmtFinType1"] = df_["BsmtFinType1"].fillna("NoBsmt")

    for col in ["LotArea", "2ndFlrSF"]:
        for df_ in (X_tr, X_va, X_te):
            df_[col] = np.log1p(df_[col].astype(float))
    
    qual_map = {'Ex':5, 'Gd':4, 'TA':3, 'Fa':2, 'Po':1, np.nan:0}
    ord_cols = ["ExterQual", "BsmtQual", "GarageQual", "GarageCond", "HeatingQC", "FireplaceQu", "BsmtCond", "KitchenQual"]
    for c in ord_cols:
        for df_ in (X_tr, X_va, X_te):
            df_[c] = df_[c].map(qual_map).astype(float)
            
    te_cols = [c for c in ["Neighborhood", "BsmtFinType1", "Foundation"] if c in X_tr.columns]
    tec = TargetEncoder(cols=te_cols, handle_unknown="value", handle_missing="value")
    X_tr[te_cols] = tec.fit_transform(X_tr[te_cols], y_tr)
    X_va[te_cols] = tec.transform(X_va[te_cols])
    X_te[te_cols] = tec.transform(X_te[te_cols])

    
    oscaler = OrdinalEncoder(handle_unknown="use_encoded_value", unknown_value=-1)
    for df_, tag in ((X_tr, "fit"), (X_va, "va"), (X_te, "te")):
        if tag == "fit":
            X_tr[["GarageType","SaleCondition", "HouseStyle", "BsmtExposure"]] = oscaler.fit_transform(X_tr[["GarageType","SaleCondition", "HouseStyle", "BsmtExposure"]])
        else:
            df_[["GarageType","SaleCondition", "HouseStyle", "BsmtExposure"]] = oscaler.transform(df_[["GarageType","SaleCondition", "HouseStyle", "BsmtExposure"]])

    for df_ in (X_tr, X_va, X_te):
        df_["TotalSF"]   = df_["TotalBsmtSF"] + df_["1stFlrSF"] + df_["2ndFlrSF"]
        df_["Age"]       = df_["YrSold"] - df_["YearBuilt"]
        df_["RemodAge"]  = df_["YrSold"] - df_["YearRemodAdd"]
        df_["TotalBath"] = df_["FullBath"] + 0.5*df_["HalfBath"] + df_["BsmtFullBath"] + 0.5*df_["BsmtHalfBath"]
        df_["OverallGrade"] = df_["OverallQual"] * df_["OverallCond"]
        df_["QualAge"]      = df_["OverallQual"] * df_["Age"]

    def add_advanced(df):
        df = df.copy()
        df["GrLivArea_per_LotArea"]  = safe_div(df["GrLivArea"], df["LotArea"])
        df["TotalSF_per_LotArea"]    = safe_div(df["TotalSF"],   df["LotArea"])
        df["GarageArea_per_GrLivArea"]= safe_div(df["GarageArea"], df["GrLivArea"])
        df["BsmtFinRatio"]           = safe_div(df["BsmtFinSF1"], df["TotalBsmtSF"])

        df["Qual_per_Age"]           = safe_div(df["OverallQual"], (df["Age"] + 1))
        df["Baths_per_GrLivArea"]    = safe_div(df["TotalBath"], df["GrLivArea"])
        df["Fireplace_per_GrLivArea"]= safe_div(df["Fireplaces"], df["GrLivArea"])

        df["HasFireplace"] = (df["Fireplaces"] > 0).astype(int)
        df["Has2ndFloor"]  = (df["2ndFlrSF"]  > 0).astype(int)

        df["HasRemodel"]   = (df["YearBuilt"] != df["YearRemodAdd"]).astype(int)

        df["Age_when_sold"]      = df["YrSold"] - df["YearBuilt"]
        df["RemodAge_when_sold"] = df["YrSold"] - df["YearRemodAdd"]
        df["Sqrt_Age"]      = np.sqrt(np.maximum(df["Age"], 0) + 1)
        df["Sqrt_RemodAge"] = np.sqrt(np.maximum(df["RemodAge"], 0) + 1)
        df["OverallQual_GrLivArea"] = df["OverallQual"] * df["GrLivArea"]
        df["TotalSF"] = df["TotalBsmtSF"] + df["1stFlrSF"] + df["2ndFlrSF"]
        df["Qual_Age"] = df["OverallQual"] * (df["YrSold"] - df["YearBuilt"])


        for c in ["GrLivArea","TotalSF","GarageArea"]:
            df[f"Log_{c}"] = np.log1p(df[c].astype(float))

        return df
    X_tr = add_advanced(X_tr)
    X_va = add_advanced(X_va)
    X_te = add_advanced(X_te)
    
    big_gar_th = X_tr["GarageArea"].median()
    for df_ in (X_tr, X_va, X_te):
        df_["BigGarage"] = (df_["GarageArea"] > big_gar_th).astype(int)
    nb_price_med = X_tr.groupby("Neighborhood")["Neighborhood"].apply(lambda s: s.index).index  # dummy
    nb_price_med = X_tr.join(np.log1p(y_tr).rename("log_price")).groupby("Neighborhood")["log_price"].median()
    nb_age_med  = X_tr.groupby("Neighborhood")["Age"].median()
    nb_qual_med = X_tr.groupby("Neighborhood")["OverallQual"].median()
    def map_nb_stats(df):
        df = df.copy()
        df["NB_Median_LogPrice"] = df["Neighborhood"].map(nb_price_med).fillna(nb_price_med.median())
        df["NB_Median_Age"]      = df["Neighborhood"].map(nb_age_med).fillna(nb_age_med.median())
        df["NB_Median_Qual"]     = df["Neighborhood"].map(nb_qual_med).fillna(nb_qual_med.median())
        return df
    X_tr = map_nb_stats(X_tr)
    X_va = map_nb_stats(X_va)
    X_te = map_nb_stats(X_te)

    scale_cols = [
        "LotArea","MSSubClass","TotalBsmtSF","1stFlrSF","GrLivArea","GarageArea",
        "TotalSF","Age","RemodAge","TotalBath",
        "GrLivArea_per_LotArea","TotalSF_per_LotArea","GarageArea_per_GrLivArea",
        "BsmtFinRatio","Qual_per_Age","Baths_per_GrLivArea","Fireplace_per_GrLivArea",
        "Age_when_sold","RemodAge_when_sold","Sqrt_Age","Sqrt_RemodAge",
        "Log_GrLivArea","Log_TotalSF","Log_GarageArea",
        "NB_Median_LogPrice","NB_Median_Age","NB_Median_Qual"
    ]
    scale_cols = [c for c in scale_cols if c in X_tr.columns]

    ss = StandardScaler()
    X_tr[scale_cols] = ss.fit_transform(X_tr[scale_cols])
    X_va[scale_cols] = ss.transform(X_va[scale_cols])
    X_te[scale_cols] = ss.transform(X_te[scale_cols])
    return X_tr, X_va, X_te
```

```python
DROPPING = [
    "PavedDrive","RoofStyle","BsmtFinType2","EnclosedPorch","MiscVal",
    "LandSlope","PoolArea","Fence","SaleType","BsmtFinSF2","LandContour",
    "ExterCond","Exterior1st","3SsnPorch","Electrical","Street","Alley",
    "Utilities","Condition2","PoolQC","RoofMatl","Heating","LowQualFinSF",
    "MiscFeature","GarageFinish","LotFrontage","WoodDeckSF","ScreenPorch",
    "MasVnrArea","MasVnrType","LotConfig","LotShape","Exterior2nd","BldgType"
]
```

```python
def get_models(random_state):
    return {
        "xgb": xgb.XGBRegressor(
            booster="gbtree", learning_rate=0.02, n_estimators=8000,
            max_depth=8, min_child_weight=1, gamma=0.1, subsample=0.8,
            colsample_bytree=0.8, reg_alpha=1.5, reg_lambda=3.0,
            max_bin=256, grow_policy="lossguide", tree_method="hist",
            device="cuda", random_state=random_state, eval_metric="rmse"
        ),
        "lgb": lgb.LGBMRegressor(
            learning_rate=0.02, n_estimators=8000, num_leaves=63,
            subsample=0.8, colsample_bytree=0.8, reg_alpha=1.5, reg_lambda=3.0,
            random_state=random_state, device="gpu"
        ),
        "cat": CatBoostRegressor(
            learning_rate=0.02, iterations=8000, depth=8,
            l2_leaf_reg=3.0, subsample=0.8, bootstrap_type="Bernoulli",
            random_seed=random_state, task_type="GPU", verbose=0
        ),
        "ridge": RidgeCV(alphas=np.logspace(-3, 3, 50)),
        "lasso": LassoCV(alphas=np.logspace(-4, 0, 50), max_iter=5000, random_state=random_state),
        "elastic": ElasticNetCV(l1_ratio=[.1, .5, .9], alphas=np.logspace(-4, 0, 30), max_iter=5000, random_state=random_state),
        "et": ExtraTreesRegressor(n_estimators=200, random_state=random_state, n_jobs=-1)
    }
```

```python
def oof(train_df, test_df, target_col="SalePrice", n_splits=5, random_state=42):
    y = train_df[target_col].copy()
    X = train_df.drop(columns=[target_col]).drop(columns=DROPPING, errors="ignore").copy()
    X_test = test_df.drop(columns=DROPPING, errors="ignore").copy()
    
    bins = pd.qcut(y, q=10, labels=False, duplicates="drop")
    kf = StratifiedKFold(n_splits, shuffle=True,
               random_state=random_state)

    models = get_models(random_state)
    oof_preds = {m: np.zeros(len(X)) for m in models}
    test_preds = {m: np.zeros(len(X_test)) for m in models}
   
    for fold, (tr_idx, va_idx) in enumerate(kf.split(X, bins), 1):
        print(f"[Fold {fold}]")
        X_tr, X_va = X.iloc[tr_idx], X.iloc[va_idx]
        y_tr, y_va = y.iloc[tr_idx], y.iloc[va_idx]

        X_tr, X_va, X_te = preprocessing(X_tr, X_va, X_test, y_tr)
        y_tr_log = np.log1p(y_tr)
        y_va_log = np.log1p(y_va)
        
        for col in X_tr.columns:
            mean_val = X_tr[col].mean()
            X_tr[col].fillna(mean_val, inplace=True)
            X_va[col].fillna(mean_val, inplace=True)
            X_te[col].fillna(mean_val, inplace=True)

        for name, model in models.items():
            model.fit(X_tr, y_tr_log)
            oof_preds[name][va_idx] = model.predict(X_va)
            test_preds[name] += model.predict(X_te) / n_splits
    oof_matrix = np.column_stack([oof_preds[m] for m in models])
    test_matrix = np.column_stack([test_preds[m] for m in models])
    
    meta = ElasticNetCV(l1_ratio=[.1, .5, .9], alphas=np.logspace(-4, 0, 30), 
                        max_iter=5000, random_state=random_state)
    meta.fit(oof_matrix, np.log1p(y))
    final_oof = meta.predict(oof_matrix)
    final_test = meta.predict(test_matrix)
    
    def rmsle(y_true, y_pred):
        return np.sqrt(np.mean((np.log1p(y_true) - np.log1p(np.expm1(y_pred)))**2))
    
    scores = {m: rmsle(y, np.expm1(oof_preds[m])) for m in models}
    scores["meta"] = rmsle(y, np.expm1(final_oof))
    
    submission = pd.DataFrame({"Id": test_df.index, target_col: np.expm1(final_test)})

    return {"oof_scores": scores, "submission": submission}
```

```python
results = oof(train_df, test_df, target_col="SalePrice", n_splits=10, random_state=42)
results["submission"].to_csv("submission_010_target.csv", index=False)
print(results["oof_scores"])
```
